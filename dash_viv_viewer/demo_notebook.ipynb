# VivViewer Demo Notebook

This notebook demonstrates the workflow for visualizing local images using the **VivViewer** high-performance component.

### Step 1: Configuration
Define your image path and output directory.

```python
import os
from dash import Dash, html
from dash_viv_viewer import VivViewer, convert_to_ome_tiff, serve_directory

# INPUT_IMAGE = "./demo_data/Visium_14_CK292.jpeg"
INPUT_IMAGE = "/Users/alex/Downloads/data20_V1_Hskin_Melanoma_Xenium_tissue_image_PYRAMIDAL.tiff"
OUTPUT_DIR = os.path.abspath("./demo_data")
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

### Step 2: Convert to OME-TIFF Pyramid
We convert the PNG to an OME-TIFF pyramid. This format is optimized for streaming large images without loading the whole file into memory.

```python
print(f"Converting {INPUT_IMAGE}...")
ome_filename = "test_pyramid.ome.tif"
ome_path = convert_to_ome_tiff(INPUT_IMAGE, output_path=os.path.join(OUTPUT_DIR, ome_filename))
print(f"Done: {ome_path}")
```

### Step 3: Start the Image Server
This launches a background thread to serve the tiles from your local directory.

```python
base_url = serve_directory(OUTPUT_DIR, port=5002)
image_url = f"{base_url}/{ome_filename}"
print(f"Image is now hosted at: {image_url}")
```

### Step 4: Display in VivViewer
We use Dash to render the viewer. Note: If you are running in a Jupyter environment, you can use `inline` mode.

```python
app = Dash(__name__)

app.layout = html.Div([
    html.H2("Interactive VivViewer"),
    VivViewer(
        id="viewer",
        image_url=image_url,
        height=600
    )
])

# Run the app
# Use mode='inline' if your Jupyter environment supports it
app.run_server(debug=True, port=8051)
```

We use Dash to render the viewer. Note: If you are running in a Jupyter environment, you can use `inline` mode.
